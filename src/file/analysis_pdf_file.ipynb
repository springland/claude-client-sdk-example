This is an example to upload a pdf file , get file id then  reference to it

deepseek API is not compatible

```python

from util.resource_path import  get_resource_path


resource_path = get_resource_path('pdfs')
resource_file = resource_path / 'cdc_84800_DS1.pdf'

```

```python
print(resource_file)
```

```python
from util.claude_client import get_claude_client_and_default_model

client , llm  = get_claude_client_and_default_model('claude')

```

```python
uploaded = client.files.upload(
    file=("document.pdf", open(resource_file, "rb"), "application/pdf"),
)

print(uploaded)
file_id = uploaded.id
print(file_id)
```

```python
response = client.messages.create(
    model=llm,
    max_tokens=1024,
    messages=[
        {
            "role": "user",
            "content": [
                {"type": "text", "text": "Please summarize this document for me."},
                {
                    "type": "document",
                    "source": {
                        "type": "file",
                        "file_id": file_id,
                    },
                },
            ],
        }
    ],
)
print(response)
```
