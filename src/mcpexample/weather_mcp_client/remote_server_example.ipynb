This one uses message api

start local http-server

Then run ngrox : ngrok http 8000 --url https://hummus-unease-dragonfly.ngrok-free.dev

Create client

```python
from util.claude_client import get_claude_client_and_default_model


client , llm_model = get_claude_client_and_default_model(profile_name="claude")
```

List tools

```python


response = client.beta.messages.create(
    model=llm_model,
    max_tokens=1000,
    messages=[{"role": "user", "content": "What tools do you have available?"}],
    mcp_servers=[
        {
            "type": "url",
            "url": "https://hummus-unease-dragonfly.ngrok-free.dev/mcp",
            "name": "weather-mcp"

        }
    ],
    tools=[{"type": "mcp_toolset", "mcp_server_name": "weather-mcp"}],
    betas=["mcp-client-2025-11-20"],
)

print( response.model_dump_json(indent=4) )

print()

for block in response.content:
    if block.type == "text":
        print(block.text)
```

Disable get_alert tool , get_forecast is enabled

```python
response = client.beta.messages.create(
    model=llm_model,
    max_tokens=1000,
    messages=[{"role": "user", "content": "What is weather for New York today and any alerts ?"}],
    mcp_servers=[
        {
            "type": "url",
            "url": "https://hummus-unease-dragonfly.ngrok-free.dev/mcp",
            "name": "weather-mcp"

        }
    ],
    tools=[
            {
                "type": "mcp_toolset",
                "mcp_server_name": "weather-mcp",
                "default_config": {"enabled": False},
                "configs": {
                    "get_forecast": {"enabled": True},
                },
            }
    ],
    betas=["mcp-client-2025-11-20"],
)

print( response.model_dump_json(indent=4) )

print()

for block in response.content:
    if block.type == "text":
        print(block.text)
```
