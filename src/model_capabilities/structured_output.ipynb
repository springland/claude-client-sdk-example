### Haiku
haiku-4-5 was trained in 2025. It does not have knowledge about 2026 trump administration
We enable web_search and web_retrieve tool to allow it search on internet


```

from util.claude_client import get_claude_client_and_default_model

client , llm = get_claude_client_and_default_model(profile_name='claude' , model_name='claude-haiku-4-5-20251001')

output_config={
    "format": {
        "type": "json_schema",
        "schema": {
            "type": "object",
            "properties": {
                "officers": {
                    "type": "array",
                    "description": "Exactly five major administration officials.",
                    "items": {
                        "type": "object",
                        "properties": {
                            "name": {"type": "string"},
                            "office": {"type": "string"},
                            "gender": {"type": "string"},
                            "date_of_birth": {
                                "type": "string",
                                "description": "Birth date in YYYY-MM-DD format.",
                            },
                        },
                        "required": [
                            "name", "office", "gender", "date_of_birth"
                        ],
                        "additionalProperties": False,
                    },
                },
            },
            "required": ["officers"],
            "additionalProperties": False,
        },
    }
}
messages = [
        {
            'role' : 'user',
            'content' : [

                {
                    'type' : 'text',
                    'text' : 'List top 5 major officers in Trump 2026 administration',
                }
            ]
        }
    ]


web_tools = [
    {"type": "web_search_20250305", "name": "web_search", "max_uses": 3},
    {"type": "web_fetch_20250910", "name": "web_fetch", "max_uses": 5},
]


message = client.messages.create(
    model = llm,
    max_tokens= 4096,
    messages = messages,
    output_config=output_config,
    tools = web_tools


)
print( message.model_dump_json(indent=4) )

print(next(block.text for block in message.content if block.type == "text"))



```
